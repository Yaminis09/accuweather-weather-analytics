```python
#imporing libraries

from pyspark.sql import functions as F
```

```python
# Hardcoded source and target table locations
SOURCE_TABLE = "samples.accuweather.historical_hourly_imperial"

TARGET_TABLE = "workspace.bronze.historical_hourly"
```

```python
# Reading source data

source_df = spark.table(SOURCE_TABLE)
```

```python
display(source_df)

source_df.printSchema()

print(source_df.count())
```

```python
# Creating bronze_df and adding ingestion timestamps
bronze_df = (
    source_df
    .withColumn(
        "ingestion_timestamp",
        F.current_timestamp()
    )
)
```

```python
# adding data to bronze table

bronze_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(TARGET_TABLE)
```

```python
bronze_table = spark.table(TARGET_TABLE)

display(bronze_table)
```
